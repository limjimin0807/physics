```python
import math    #수학 함수를 사용하기 위해 라이브러리 추가
```

```python
angel_degree = 30
angel_radian = angel_degree * math.pi / 180
print("angel_radian :", angel_radian)
```

```python
math.cos(angel_radian)
```

```python
math.sin(angel_radian)
```

```python
math.tan(angel_radian)
```

```python
x, y = 1.0, 2.0
```

```python
r = math.sqrt(x**2 + y**2)
theta_raidan = math.atan2(y, x)

print(r, theta_raidan)
```
